```python
import math
import seaborn as sns
import numpy as np
import scanpy as sc
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt
from abc_atlas_access.abc_atlas_cache.abc_project_cache import AbcProjectCache
```

```python
# Selecting the brain region
select_region = "Microglia"
```

```python
# Loading AnnData object
base_path = Path("/data/scRNA/ABCA/AIBS/AWS/expression_matrices/WMB-10Xv3/20230630/")
expr_path = base_path / f"WMB-10Xv3-{select_region}-raw-wmeta.h5ad"
adata = sc.read_h5ad(expr_path)
adata
```

```python
# Loading Module Eigengenes dataframe
base_path = Path("/data/scRNA/ABCA/AIBS/AWS/expression_matrices/WMB-10Xv3/20230630/outputs")
hubs_df = pd.read_csv(base_path / f"WMB-10Xv3-{select_region}-raw-mc-wgcna-hubs.csv", index_col="Unnamed: 0")
hubs_df.head()
```

```python
# Map gene names to gene symbols using adata.var["gene_symbol"]
gene_symbol_mapping = adata.var["gene_symbol"].to_dict()

# Add a new column 'gene_symbol' to hubs_df
hubs_df["gene_symbol"] = hubs_df["gene_name"].map(gene_symbol_mapping)

# Display the updated dataframe
hubs_df.head()
```

```python
# Loading Module Eigengenes dataframe
base_path = Path("/data/scRNA/ABCA/AIBS/AWS/expression_matrices/WMB-10Xv3/20230630/outputs")
hubs_df.to_csv(base_path / f"WMB-10Xv3-{select_region}-raw-mc-wgcna-hubs-mod.csv")
```
